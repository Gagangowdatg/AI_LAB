```python
print("GAGAN GOWDA T G")
print("1BM24CS098")
print("AI_LAB-H_WEEK02")
# 8-Puzzle Problem Solver using Depth-First Search (DFS)
import sys

# Increase recursion depth limit just in case for deep DFS branches
sys.setrecursionlimit(100000)

class PuzzleState:
    def __init__(self, board, parent=None, move="", depth=0):
        self.board = board       # 1D tuple representing the 3x3 board
        self.parent = parent     # Pointer to the parent node to reconstruct the path
        self.move = move         # The move that led to this state (Up, Down, Left, Right)
        self.depth = depth       # Current depth level in the search tree

    def __eq__(self, other):
        return self.board == other.board

    def __hash__(self):
        return hash(self.board)

def get_possible_moves(state):
    """Generates all valid next states by moving the blank tile (0)."""
    moves = []
    board = list(state.board)
    blank_idx = board.index(0)
    row, col = blank_idx // 3, blank_idx % 3

    # Define directional offsets: (row_change, col_change, move_name)
    directions = [(-1, 0, "Up"), (1, 0, "Down"), (0, -1, "Left"), (0, 1, "Right")]

    for dr, dc, move_name in directions:
        new_row, new_col = row + dr, col + dc
        if 0 <= new_row < 3 and 0 <= new_col < 3:
            new_idx = new_row * 3 + new_col
            # Clone and swap tiles
            new_board = board[:]
            new_board[blank_idx], new_board[new_idx] = new_board[new_idx], new_board[blank_idx]
            moves.append(PuzzleState(tuple(new_board), state, move_name, state.depth + 1))
    return moves

def solve_dfs(initial_board, goal_board, depth_limit=50):
    """Solves the 8-puzzle using Depth-Limited DFS to handle infinite loops or deep paths."""
    start_state = PuzzleState(tuple(initial_board))
    goal_state = PuzzleState(tuple(goal_board))

    # Using an explicit Stack for DFS: stores tuples of (state)
    stack = [start_state]
    # Visited tracking to avoid cyclical paths
    visited = set()

    nodes_expanded = 0

    while stack:
        current_state = stack.pop()
        nodes_expanded += 1

        if current_state.board == goal_state.board:
            # Reconstruct path
            path = []
            curr = current_state
            while curr.parent:
                path.append((curr.move, curr.board))
                curr = curr.parent
            path.reverse()
            return path, nodes_expanded

        if current_state not in visited and current_state.depth < depth_limit:
            visited.add(current_state)
            # Add neighbors to stack (DFS behavior)
            for neighbor in get_possible_moves(current_state):
                if neighbor not in visited:
                    stack.append(neighbor)

    return None, nodes_expanded

def print_board(board_tuple):
    for i in range(0, 9, 3):
        print(f" {board_tuple[i]} {board_tuple[i+1]} {board_tuple[i+2]} ")
    print("-" * 11)

# --- Interactive Input and Execution Setup ---
print("--- Enter the 8-Puzzle Initial State (use 0 for blank space) ---")
print("Enter elements row by row separated by spaces (e.g., 1 2 3):")
initial = []
for i in range(3):
    row = list(map(int, input(f"Row {i+1}: ").split()))
    initial.extend(row)

print("\n--- Enter the 8-Puzzle Goal State ---")
goal = []
for i in range(3):
    row = list(map(int, input(f"Row {i+1}: ").split()))
    goal.extend(row)

print("\nInitial State:")
print_board(initial)
print("Goal State:")
print_board(goal)

print("\nSolving using DFS... Please wait...")
# Depth limit safety to prevent colab crashes since pure DFS can wander infinitely
solution_path, total_nodes = solve_dfs(initial, goal, depth_limit=30)

if solution_path is not None:
    print(f"\n🎉 Goal Reached in {len(solution_path)} steps!")
    print(f"Total nodes processed: {total_nodes}\n")
    print("Step-by-Step Path:")
    print_board(initial)
    for move, board in solution_path:
        print(f"Move: {move}")
        print_board(board)
else:
    print(f"\n❌ No solution found within depth limit or puzzle is unsolvable.")
    print(f"Total nodes evaluated: {total_nodes}")
```

```python
print("GAGAN GOWDA T G")
print("1BM24CS098")
print("AI_LAB-H")
import time

# --- FIXED MOVES DICTIONARY ---
# The second element in the tuple MUST be the relative index change:
# Moving Up: -3 | Moving Down: +3 | Moving Left: -1 | Moving Right: +1
MOVES = {
    0: [('Down', 3), ('Right', 1)],
    1: [('Down', 3), ('Left', -1), ('Right', 1)],
    2: [('Down', 3), ('Left', -1)],
    3: [('Up', -3), ('Down', 3), ('Right', 1)],
    4: [('Up', -3), ('Down', 3), ('Left', -1), ('Right', 1)],
    5: [('Up', -3), ('Down', 3), ('Left', -1)],
    6: [('Up', -3), ('Right', 1)],
    7: [('Up', -3), ('Left', -1), ('Right', 1)],
    8: [('Up', -3), ('Left', -1)]
}

def get_successors(state):
    """Generates next valid states and actions from the current state."""
    successors = []
    blank_idx = state.index(0)

    for action, shift in MOVES[blank_idx]:
        new_idx = blank_idx + shift

        # Swap blank space with the target tile
        new_state = list(state)
        new_state[blank_idx], new_state[new_idx] = new_state[new_idx], new_state[blank_idx]
        successors.append((tuple(new_state), action))

    return successors

def depth_limited_search(state, goal, depth, path, actions):
    """Performs Depth-First Search up to a specific depth limit."""
    if state == goal:
        return actions
    if depth <= 0:
        return "cutoff"

    cutoff_occurred = False
    for next_state, action in get_successors(state):
        # Avoid cyclic paths within the current branch
        if next_state not in path:
            path.add(next_state)
            result = depth_limited_search(next_state, goal, depth - 1, path, actions + [action])

            if result == "cutoff":
                cutoff_occurred = True
            elif result is not None:
                return result
            path.remove(next_state)  # Backtrack

    return "cutoff" if cutoff_occurred else None

def iterative_deepening_search(initial_state, goal_state, max_depth=50):
    """Iteratively deepens the DLS depth until a solution is found."""
    print("Searching for a solution...")
    start_time = time.time()

    for depth in range(max_depth):
        path = {initial_state}
        result = depth_limited_search(initial_state, goal_state, depth, path, [])
        if result != "cutoff":
            end_time = time.time()
            if result is not None:
                print(f"🎉 Solution found at depth: {depth}")
                print(f"⏱️ Time taken: {end_time - start_time:.4f} seconds")
                return result
            else:
                print("❌ No solution exists.")
                return None

    print("⚠️ Reached max depth limit without finding a solution.")
    return None

def print_board(state):
    """Utility function to print the 3x3 board cleanly."""
    for i in range(0, 9, 3):
        print(f"[ {state[i]} {state[i+1]} {state[i+2]} ]")
    print()

# --- USER INPUT & EXECUTION ---
print("--- Enter States as 9 space-separated numbers (0-8) ---")
print("Example Easy Initial State: 1 2 3 4 0 5 6 7 8")
print("Example Goal State        : 1 2 3 4 5 6 7 8 0\n")

try:
    init_input = input("Enter Initial State: ").strip().split()
    goal_input = input("Enter Goal State: ").strip().split()

    initial_state = tuple(map(int, init_input))
    goal_state = tuple(map(int, goal_input))

    if len(initial_state) == 9 and len(goal_state) == 9:
        print("\nInitial Board:")
        print_board(initial_state)
        print("Goal Board:")
        print_board(goal_state)

        solution = iterative_deepening_search(initial_state, goal_state)

        if solution:
            print(f"Steps to solve ({len(solution)} moves): {', '.join(solution)}")
    else:
        print("Error: Please make sure you enter exactly 9 digits.")
except ValueError:
    print("Error: Invalid input. Please enter numbers only.")
```
